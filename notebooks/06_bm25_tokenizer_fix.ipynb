```python
from src.utils import (load_swefaq, load_swedn, build_qrels_dict, build_bm25,
                       tokenize, evaluate_retrieval)

swefaq_corpus, swefaq_queries, swefaq_qrels = load_swefaq()
swedn_corpus, swedn_queries, swedn_qrels = load_swedn()
swefaq_qrels_dict = build_qrels_dict(swefaq_qrels)
swedn_qrels_dict = build_qrels_dict(swedn_qrels)

def make_bm25_search(corpus):
    bm25, ids, texts = build_bm25(corpus)
    def search(q, k=10):
        scores = bm25.get_scores(tokenize(q))
        top = scores.argsort()[::-1][:k]
        return [ids[i] for i in top]
    return search

swefaq_res = evaluate_retrieval(make_bm25_search(swefaq_corpus), swefaq_queries, swefaq_qrels_dict)
swedn_res  = evaluate_retrieval(make_bm25_search(swedn_corpus),  swedn_queries,  swedn_qrels_dict)
print("SweFAQ BM25 (new):", swefaq_res)
print("Swedn  BM25 (new):", swedn_res)
```
